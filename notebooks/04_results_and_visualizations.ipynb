## <span style="color:#FF007F"> 2. Compare F1 scores </span>  

<a href="#top">Top &#129033;</a>

```python

f1_scores = [0.0976, 0.1541, 0.2129, 0.4578, 0.1153]

models = ['Baseline Model', 'Improved Baseline Model', 'Complex CNN Model', 'ResNet50', 'VGG-16']
```

```python
sorted_indices = sorted(range(len(f1_scores)), key=lambda k: f1_scores[k], reverse=True)
sorted_models = [models[i] for i in sorted_indices]
sorted_f1_scores = [f1_scores[i] for i in sorted_indices]

plt.figure(figsize=(15, 6))
bars = plt.bar(sorted_models, sorted_f1_scores, color="pink")

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval/2, f'{yval:.3f}', ha='center', va='center', color='black', fontweight='bold')

plt.xlabel('Model Names')
plt.ylabel('F1 Scores')
plt.title('Comparison of F1 Scores')
plt.ylim(0, 0.35) 
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

## <span style="color:#FF007F"> 3. Compare Accuracy </span>  

<a href="#top">Top &#129033;</a>

```python
accuracy_scores = [0.1291, 0.1919, 0.2485, 0.4898, 0.1565]
```

```python
sorted_acc_indices = sorted(range(len(accuracy_scores)), key=lambda k: accuracy_scores[k], reverse=True)
sorted_acc_models = [models[i] for i in sorted_acc_indices]
sorted_accuracy_scores = [accuracy_scores[i] for i in sorted_acc_indices]

plt.figure(figsize=(15, 6))
acc_bars = plt.bar(sorted_acc_models, sorted_accuracy_scores, color="pink")

for bar in acc_bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval/2, f'{yval:.3f}', ha='center', va='center', color='black', fontweight='bold')

plt.xlabel('Model Names')
plt.ylabel('Accuracy')
plt.title('Comparison of Accuracy')
plt.ylim(0, 0.35) 
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```
